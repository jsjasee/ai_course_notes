## Now let's create 3 pretend functions..


Recap how tools work:

1. You come up with some janky Json that describes to a model what functionality / capabilities it has
2. When we prompt a model, we tell the model it has access to these tools via the json and if they want to call any of these tools
3. If yes, then there's an if statement -> if it wants to call any of these tools, we will do so and return the results

### OK now a big block of JSON

Note: if we use an agent framework, this json below is written for us


```python
scan_function = {
    "name": "scan_the_internet_for_bargains",
    "description": "Returns top bargains scraped from the internet along with the price each item is being offered for",
    "parameters": {
        "type": "object",
        "properties": {},
        "required": [],
        "additionalProperties": False,
    },
}

estimate_function = {
    "name": "estimate_true_value",
    "description": "Given the description of an item, estimate how much it is actually worth",
    "parameters": {
        "type": "object",
        "properties": {
            "description": {
                "type": "string",
                "description": "The description of the item to be estimated",
            },
        },
        "required": ["description"],
        "additionalProperties": False,
    },
}

notify_function = {
    "name": "notify_user_of_deal",
    "description": "Send the user a push notification about the single most compelling deal; only call this one time",
    "parameters": {
        "type": "object",
        "properties": {
            "description": {
                "type": "string",
                "description": "The description of the item itself scraped from the internet",
            },
            "deal_price": {
                "type": "number",
                "description": "The price offered by this deal scraped from the internet",
            },
            "estimated_true_value": {
                "type": "number",
                "description": "The estimated actual value that this is worth",
            },
            "url": {
                "type": "string",
                "description": "The URL of this deal as scraped from the internet",
            },
        },
        "required": ["description", "deal_price", "estimated_true_value", "url"],
        "additionalProperties": False,
    },
}
```

```python
tools = [
    {"type": "function", "function": scan_function},
    {"type": "function", "function": estimate_function},
    {"type": "function", "function": notify_function},
]
```

```python
tools  # this json prompt is fed into the system prompt when the llm is called
```

```python
def handle_tool_call(message):
    """
    Actually call the tools associated with this message
    """
    results = []
    for tool_call in message.tool_calls:
        tool_name = tool_call.function.name
        arguments = json.loads(tool_call.function.arguments)
        tool = globals().get(
            tool_name
        )  # collect the function from the global namespace, from the name of the tool
        # for this to work, the name of the tool, eg. 'name': 'scan_the_internet_for_bargains', must be THE SAME AS THE FUNCTION NAME (we are tranlating a string into a function then calling it dynamically)

        # an alternative is just:
        # if tool_name == "scan_the_internet_for_bargains":
        #     scan_the_internet_for_bargains()

        result = (
            tool(**arguments) if tool else {}
        )  # call the tool describe from the response
        results.append(
            {
                "role": "tool",
                "content": json.dumps(result),
                "tool_call_id": tool_call.id,
            }
        )
    return results
```

```python
system_message = "You find great deals on bargain products using your tools, and notify the user of the best bargain."
user_message = """
First, use your tool to scan the internet for bargain deals. Then for each deal, use your tool to estimate its true value.
Then pick the single most compelling deal where the price is much lower than the estimated true value, and use your tool to notify the user.
Then just reply OK to indicate success.
"""
messages = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_message},
]
```

```python
messages
```

```python
done = False
while not done:  # this is a while loop - lets agent keep going until it's done
    response = openai.chat.completions.create(
        model=MODEL, messages=messages, tools=tools
    )
    if response.choices[0].finish_reason == "tool_calls":
        message = response.choices[0].message
        results = handle_tool_call(message)
        messages.append(message)
        messages.extend(results)
    else:
        # if the task is done, it won't call anymore tools, that's how we know its done.
        done = True
response.choices[0].message.content
```
